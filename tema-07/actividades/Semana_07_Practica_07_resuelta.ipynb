```python
#@title Solución

def e2_ingreso_dato() -> float:
  """ Ingreso de datos.

  Returns:
    float: Retorna el dato ingresado
  """
  while (valor:=input(f'Ingrese un número flotante:')):
    try:
      return float(valor)
    except ValueError:
      print(f'Debe ingresar un número entero o flotante. Usted ingreso {valor}')

e2_ingreso_dato()
```

# Ejercicio 3

Tomar la función definida y valida que el valor ingresar sea un número y que siempre retorne un valor flotante.

Está función solo admite números, entonces si pasamos un caracter no válido tira un error.

`Utilizar assert`

```python
def e3_ingreso_dato() -> float:
  """ Ingreso de datos.

  Returns:
    float: Retorna el dato ingresado
  """
  return float(input(f'Ingrese un número flotante:'))

e3_ingreso_dato()
```

```python
#@title Solución
def e3_validar(valor:float) -> bool:
  """ Función que valida si es un flotante.

  Args:
    valor (float): Contiene el número.

  Returns:
    bool: Retorna True si es un flotante.
  """
  try:
    float(valor)
    return True
  except ValueError:
    return False

def e3_ingreso_dato() -> float:
  """ Ingreso de datos.

  Returns:
    float: Retorna el dato ingresado
  """
  while (valor:=input(f'Ingrese un número flotante:')):
    try:
      assert e3_validar(valor=valor), f'Debe ingresar un número entero o flotante. Usted ingreso {valor}'
      return float(valor)
    except AssertionError as e:
      print(e)

e3_ingreso_dato()
```

# Ejercicio 4

Arregla las siguientes líneas de código para que no lanzen una excepción y ejecute el código de forma correcta.

```python

def e4_cargar_datos_diccionario():
  """
  Carga datos por consola y los guarda en un diccionario.

  Args:
    No hay argumentos.

  Returns:
    Un diccionario con los datos introducidos por el usuario.
  """
  diccionario = {}
  while True:
    clave = input('Introduzca la clave (o "salir" para terminar): ')
    if clave.lower() == 'salir':
      break
    valor = int(input('Introduzca el valor entero: '))
    diccionario[clave] = valor
  return diccionario

def e4_mostrar_datos_diccionario(diccionario:dict, titulo:str=None, ordenar_por_clave:bool=False):
  """
  Muestra los datos de un diccionario de forma elegante.

  Args:
    diccionario (dict): El diccionario con los datos a mostrar.
    titulo (str): El título a mostrar encima del diccionario (opcional).
    ordenar_por_clave (bool): Indica si se deben ordenar las claves del diccionario antes de mostrarlas (opcional).

  Returns:
    No hay retorno.
  """
  if titulo is not None:
    print(f'\n{titulo}\n')

  # Se ordenan las claves del diccionario si se ha indicado
  if ordenar_por_clave:
    claves = sorted(diccionario.keys())
  else:
    claves = list(diccionario.keys())

  for clave in claves:
    valor = diccionario[clave]
    print(f'  {clave}: ' + valor)

# Ejemplo de uso
diccionario = e4_cargar_datos_diccionario()

e4_mostrar_datos_diccionario(diccionario=diccionario,
                             titulo="Datos del diccionario")
```
