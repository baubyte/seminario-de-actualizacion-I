 Leer un archivo de texto y mostrar su contenido por pantalla.

```python
#@title Solución

# nombre del archivo
e1_nombre_archivo:str = '/content/sample_data/README.md'

def leer_archivo_texto(nombre_archivo:str):
  """ Lee un archivo de texto y lo imprime.

  Args:
    nombre_archivo (str): Contiene el nombre del archivo.
  """
  try:
    with open(nombre_archivo, 'r') as archivo:
      print(archivo.read())
  except FileNotFoundError:
    print("El archivo no se encontró.")

# Ejemplo de uso
leer_archivo_texto(nombre_archivo=e1_nombre_archivo)
```

# Ejercicio 3

Leer un archivo JSON y mostrar los datos en un árbol.


```python
# nombre del archivo
e3_nombre_archivo:str = '/content/sample_data/anscombe.json'
```

```python
#@title Solución

# nombre del archivo
e3_nombre_archivo:str = '/content/sample_data/anscombe.json'

import json
from typing import Any

def imprimir_json(data:dict|list|Any, nivel:int=0):
    """ Imprime una línea.

    Args:
        data (dict|list|Any): Contiene la data a imprimir.
        nivel (int): Contiene el nivel hacia la derecha a imprimir. Defautls to 0.
    """
    if isinstance(data, dict):
        for key, value in data.items():
            if isinstance(value, list) or isinstance(value, dict):
              print('\t' * nivel + str(key))
              imprimir_json(value, nivel + 1)
            else:
              print('\t' * nivel + str(key) + ": " + str(value))
    elif isinstance(data, list):
        for item in data:
            imprimir_json(item, nivel + 1)
            print()
    else:
        print('\t' * nivel + str(data))

def leer_archivo_json(nombre_archivo:str):
    """ Lee un archivo json.

    Args:
      nombre_archivo (str): Contiene el nombre del archivo.
    """
    try:
        with open(nombre_archivo, 'r') as archivo:
            contenido = archivo.read()
            datos_json = json.loads(contenido)
            imprimir_json(datos_json)
    except FileNotFoundError:
        print("El archivo no se encontró.")
    except json.decoder.JSONDecodeError:
        print("El archivo JSON está mal formado.")

# Ejemplo de uso
leer_archivo_json(nombre_archivo=e3_nombre_archivo)
```

# Ejercicio 5
Escribir un archivo CSV con una lista de datos.

Comprobarlo con la solución desarrollada en el ejercicio 2.

```python
# nombre del archivo
e5_nombre_archivo:str = '/content/ejercicio_5.csv'

# Ejemplo de uso
datos = [
    ["Nombre", "Edad", "Ciudad"],
    ["Juan", 30, "Madrid"],
    ["María", 25, "Barcelona"],
    ["Pedro", 35, "Sevilla"]
]
```

```python
#@title Solución

# nombre del archivo
e5_nombre_archivo:str = '/content/ejercicio_5.csv'

# Ejemplo de uso
datos = [
    ["Nombre", "Edad", "Ciudad"],
    ["Juan", 30, "Madrid"],
    ["María", 25, "Barcelona"],
    ["Pedro", 35, "Sevilla"]
]

import csv

def escribir_archivo_csv(nombre_archivo:str, datos:list[list[str]]):
    """ Escribe un archivo de texto csv.

    Args:
      nombre_archivo (str): Contiene el nombre del archivo.
      datos (list[list[str]]): Contiene los datos a escribir.
    """
    try:
        with open(nombre_archivo, 'w', newline='') as archivo:
            escritor_csv = csv.writer(archivo)
            for fila in datos:
                escritor_csv.writerow(fila)
        print("Se ha escrito el archivo CSV correctamente.")
    except Exception as e:
        print(f"Ocurrió un error al escribir el archivo CSV: {e}")


escribir_archivo_csv(nombre_archivo=e5_nombre_archivo, datos=datos)

leer_archivo_csv(nombre_archivo=e5_nombre_archivo)
```

# Ejercicio 6
Escribir un archivo JSON con un diccionario de datos.

Comprobarlo con la solución desarrollada en el ejercicio 3.

```python
# nombre del archivo
e6_nombre_archivo:str = '/content/ejercicio_6.json'
# Ejemplo de uso
datos = {
    "nombre": "Juan",
    "edad": 30,
    "ciudad": "Madrid"
}
```

```python
#@title Solución

# nombre del archivo
e6_nombre_archivo:str = '/content/ejercicio_6.json'
# Ejemplo de uso
datos = {
    "nombre": "Juan",
    "edad": 30,
    "ciudad": "Madrid"
}

import json

def escribir_archivo_json(nombre_archivo:str, datos:dict):
    """ Escribe un archivo de texto json.

    Args:
      nombre_archivo (str): Contiene el nombre del archivo.
      datos (dict): Contiene los datos a escribir.
    """
    try:
        with open(nombre_archivo, 'w') as archivo:
            json.dump(datos, archivo, indent=4)
        print("Se ha escrito el archivo JSON correctamente.")
    except Exception as e:
        print(f"Ocurrió un error al escribir el archivo JSON: {e}")

escribir_archivo_json(nombre_archivo=e6_nombre_archivo, datos=datos)

leer_archivo_json(nombre_archivo=e6_nombre_archivo)
```

# Ejercicio 7
Buscar una cadena en un archivo de texto.


```python
# nombre del archivo
e7_nombre_archivo:str = '/content/sample_data/README.md'

cadenas_a_buscar:list[str] = ['california_housing_data', 'datetime', 'ist_']
```

```python
#@title Solución

# nombre del archivo
e7_nombre_archivo:str = '/content/sample_data/README.md'

cadenas_a_buscar:list[str] = ['california_housing_data', 'datetime', 'ist_']

def buscar_cadena_en_archivo(nombre_archivo:str, cadena_buscar:str):
    """ Busca una cadena en un arhcivo.

    Args:
      nombre_archivo (str): Contiene el nombre del archivo.
      cadena_buscar (str): Contiene la cadena a buscar.
    """
    try:
        with open(nombre_archivo, 'r') as archivo:
            contenido = archivo.read()
            if cadena_buscar in contenido:
                print(f"La cadena '{cadena_buscar}' fue encontrada en el archivo.")
            else:
                print(f"La cadena '{cadena_buscar}' no fue encontrada en el archivo.")
    except FileNotFoundError:
        print("El archivo no se encontró.")
    except Exception as e:
        print(f"Ocurrió un error: {e}")

for cadena in cadenas_a_buscar:
  buscar_cadena_en_archivo(e7_nombre_archivo, cadena_buscar=cadena)
```
