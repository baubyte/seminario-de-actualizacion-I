## bool
Al igual que en otros lenguajes de programación, en Python existe el tipo bool o booleano. Es un tipo de dato que permite almacenar dos valores `True` (verdadero o 1) o `False` (falso o 0).

Puedes obtener más ayuda con el comando help.

```python
help(bool)
```

## Declarar variable bool
Se puede declarar una variable booleana de la siguiente manera.

```python
x:bool = True
y:bool = False
```

## Evaluar expresiones
Un valor booleano también puede ser el resultado de evaluar una expresión. Ciertos operadores como el mayor que, menor que o igual que devuelven un valor bool.

```python
print('La comparación 1 > 0 es: ', 1 > 0)
print('La comparación 1 <= 0 es: ', 1 <= 0)
print('La comparación 9 == 9 es: ', 9 == 9)
```


## Tablas de verdades

Existen operadores específicos que me permiten operar en booleanos, para cada uno de estos existe lo que se llama la tabla de verdad.

### Operador NOT (negación)

Este operador se utiliza anteponiendo la palabra `not` adelante de un tipo booleano, con lo cual la siguiente tabla se puede escribir como `not bool_` donde la variable `bool_` toma los valores de cada fila.

| bool_ | resultado |
| ---   | --- |
| True  | False |
| False | True |

### Operador AND (intersección)

Este operador se utiliza poniendo entre dos tipos booleanos la palabra `and`, con lo cual la siguiente tabla se puede escribir como ` bool_1 and bool_2` donde las variables `bool_X` toma los valores de cada fila.

| bool_1 | bool_2 | resultado |
| --- | --- | --- |
| True | True | True |
| True | False | False |
| False | False | False |
| False | True | False

### Operador OR (unión)

Este operador se utiliza poniendo entre dos tipos booleanos la palabra `or`, con lo cual la siguiente tabla se puede escribir como ` bool_1 or bool_2` donde las variables `bool_X` toma los valores de cada fila.

| bool_1 | bool_2 | resultado |
| --- | --- | --- |
| True | True | True |
| True | False | True |
| False | False | False |
| False | True | True |

## Función bool
También es posible convertir un determinado valor a bool usando la función `bool`. Todo valor distinto de 0 o vacio en el caso de los iterables se transforma en `False` mientras que cualquier otro valor es `True`.

```python
print('La transformación de 10 es: ', bool(10))
print('La transformación de -10 es: ', bool(-10))
print('La transformación de Hola es: ', bool("Hola"))
print('La transformación de 0.1 es: ', bool(0.1))
print('La transformación de [] es: ', bool([]))
print('La transformación de "" es: ', bool(""))
print('La transformación de 0 es: ', bool(0))
```

## Uso con if
Los condicionales if evalúan una condición que es un valor bool.

```python
a:bool = 1
b:bool = 2
if b > a:
    print("b es mayor que a")
else:
    print("a es mayor que b")
```

La expresión que va después del if es siempre evaluada como un booleano.



```python
if True:
    print("Es True")
```

## Bool como subclase de int
Es importante notar que aunque estemos listando el tipo bool como si fuese un tipo más, es en realidad una subclase del int (lo veremos más adelante). De hecho lo puedes comprobar de la siguiente manera.

```python
print('El valor True es una instancia de un entero?', isinstance(True, int))
print('El valor bool es una subclase de un entero?',  issubclass(bool, int))
```
