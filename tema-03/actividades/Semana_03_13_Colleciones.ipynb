# Ejercicio 3

Determinar si un triángulo es rectángulo a partir de sus longitudes de lado. El usuario debe ingresar los 3 lados.
Debemos aplicar el teorema de Pitágoras.

$a^2 + b^2 = h^2$

Donde a y b son los lados y h es la hipotenusa.
