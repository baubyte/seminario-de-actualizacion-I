# Básicas

## Ejercicio 1
Crear una clase Persona con los atributos nombre, edad y dni. Implementar un método \_\_init\_\_ para inicializar los atributos y un método mostrar_datos para imprimir la información de la persona.


## Ejercicio 2
Crear una clase Coche con los atributos marca, modelo y matricula. Implementar un método \_\_init\_\_ para inicializar los atributos y un método conducir que imprima un mensaje.

## Ejercicio 3
Crear una clase Punto con los atributos x e y. Implementar un método \_\_init\_\_ para inicializar los atributos y un método distancia que calcule la distancia a otro punto.

# Herencia

## Ejercicio 4
Crear una clase Vehiculo con los atributos nombre y potencia. Implementar un método \_\_init\_\_ para inicializar los atributos y un método cantidad\_ruedas que imprima un mensaje genérico.

## Ejercicio 5
Crear una clase Auto que herede de Vehiculo e implemente un método cantidad\_ruedas hablar que imprima un 4.

## Ejercicio 6
Crear una clase Moto que herede de Vehiculo e implemente un método cantidad\ruedas que imprima un 2.

# Polimorfismo

## Ejercicio 7
Crear una clase Figura con un método abstracto calcular_area.

## Ejercicio 8
Crear una clase Cuadrado que herede de Figura e implemente el método calcular_area para calcular el área de un cuadrado.

## Ejercicio 9
Crear una clase Triangulo que herede de Figura e implemente el método calcular_area para calcular el área de un triángulo.

## Ejercicio 10
Escribir un programa que cree una lista de figuras y calcule el área total de la lista.

# Encpsulamiento

## Ejercicio 11

## Ejercicio 12
Implementar un setter para el atributo dni que valide que el valor introducido sea un DNI válido.

## Ejercicio 13
Implementar un getter para el atributo saldo que imprima un mensaje antes de devolver el valor.
