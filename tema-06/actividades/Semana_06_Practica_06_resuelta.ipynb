## Ejercicio 3
Crear una clase Punto con los atributos x e y. Implementar un método \_\_init\_\_ para inicializar los atributos y un método distancia que calcule la distancia a otro punto.

```python
#@title Solución

from __future__ import annotations

class Punto:
  """ Clase Punto """

  def __init__(self, x:float, y:float):
    """ Constructor

    Args:
      x (float): Coordenada X.
      y (float): Coordenada Y.
    """
    self.x:float = x
    self.y:float = y

  def __str__(self):
    return f'({self.x},{self.y})'

  # para usar la anotación de Punto utilizo el 'from __future__ import annotations'
  # el cual ignora que la clase Punto todavía no esta definida por completo
  ## OJO!: EL future annotatios TAMBIÉN SE UTILIZA PARA LA MALA PRÁCTICA DE
  ## ALGUNOS PROGRAMADORES DE ANIDAR MODULOS DE FORMA RECURSICA.
  def distancia(self, punto:Punto):
    """ Calcula la distancia.

    Args:
      punto (Punto): Segundo punto para calcular la distancia.
    """
    distancia:float=pow((punto.x**2-self.x**2) + (punto.y**2-self.y**2), .5)
    print(f'La distancia entre {self} y {punto} es: {distancia:.2f}')

  # aqui la anotación de punto utilizo comilla simple para que no tire un
  # error porque clase Punto todavía no esta definida por completo
  def distancia_2(self, punto:'Punto'):
    """ Calcula la distancia.

    Args:
      punto (Punto): Segundo punto para calcular la distancia.
    """
    distancia:float=pow((punto.x**2-self.x**2) + (punto.y**2-self.y**2), .5)
    print(f'La distancia 2 entre {self} y {punto} es: {distancia:.2f}')

punto_1 = Punto(1,2)
punto_2 = Punto(2,2)

punto_1.distancia(punto=punto_2)
punto_1.distancia_2(punto=punto_2)
```

# Polimorfismo

## Ejercicio 7
Crear una clase Figura con un método abstracto calcular_area.

```python
#@title Solución

from abc import ABC, abstractmethod

class Figura(ABC):
    """ Clase abstracta Figura.
    """

    @abstractmethod
    def calcular_area(self) -> float:
        """ Calcula el área

        Returns:
            float: Returna el área.
        """
        pass
```

## Ejercicio 8
Crear una clase Cuadrado que herede de Figura e implemente el método calcular_area para calcular el área de un cuadrado.

```python
#@title Solución

class Cuadrado(Figura):
    """ Clase Cuadrado.
    """

    def __init__(self, lado:float):
        """ Constructor

        Args:
          lado (float): Medida del lado.
        """
        self.lado:float = lado

    def calcular_area(self) -> float:
        """ Calcula el área

        Returns:
            float: Returna el área.
        """
        return self.lado ** 2

cuadrado = Cuadrado(lado=3)
print(f'El área de un cuadado de lado igual a {cuadrado.lado} es : {cuadrado.calcular_area()}')
```

## Ejercicio 9
Crear una clase Triangulo que herede de Figura e implemente el método calcular_area para calcular el área de un triángulo.

```python
#@title Solución

class Triangulo(Figura):
    def __init__(self, base:float, altura:float):
        """ Constructor

        Args:
          base (float): Medida de la base.
          altura (float): Medida de la altura.
        """
        self.base:float = base
        self.altura:float = altura

    def calcular_area(self) -> float:
        """ Calcula el área

        Returns:
            float: Returna el área.
        """
        return (self.base * self.altura) / 2


triangulo = Triangulo(base=3, altura=8)
print(f'El área de un triánguloa de base igual a {triangulo.base} y de altura igual a {triangulo.altura} es : {triangulo.calcular_area()}')
```

## Ejercicio 10
Escribir un programa que cree una lista de figuras y calcule el área total de la lista.

```python
#@title Solución

def calcular_area_total(figuras:list[Figura]) -> float:
    """ Cacula el área total de las figuras

    Args:
        figuras (list[Figuras]): Lista de figuras.

    Returns:
        float: Retorna la suma de las áreas de cada figura.
    """
    return sum(map(lambda f: f.calcular_area(), figuras))

# Crear una lista de figuras
figuras = [Cuadrado(4), Triangulo(3, 5), Cuadrado(2), Triangulo(4, 6)]

# Calcular el área total de la lista de figuras
area_total = calcular_area_total(figuras)
print("El área total de la lista de figuras es:", area_total)
```

## Ejercicio 11
Crear una clase CuentaBancaria con los atributos nombre, saldo y dni. Implementar métodos para depositar y retirar dinero, y para obtener el saldo actual.

```python
#@title Solución

class CuentaBancaria:
    """ Clase CuentaBancaria
    """
    def __init__(self, nombre:str, saldo:float, dni:int):
        """ Constructor

        Args:
            nombre (str): Nombre de la cuenta.
            saldo (float): Saldo de la cuenta.
            dni (int): Dni de la persona.
        """
        self.__nombre:str = nombre
        self.__saldo:float = saldo
        self.__dni:int = dni

    def depositar(self, cantidad:float):
        """ Agrega dinero a la cuenta bancaria.

        Args:
            cantida (float): Cantidad de dinero a agregar.
        """
        self.__saldo += cantidad
        print(f"Se han depositado {cantidad} unidades. Saldo actual: {self.__saldo}")

    def retirar(self, cantidad:float):
        """ Retirar dinero a la cuenta bancaria.

        Args:
            cantida (float): Cantidad de dinero a retirar.
        """
        if cantidad <= self.__saldo:
            self.__saldo -= cantidad
            print(f"Se han retirado {cantidad} unidades. Saldo actual: {self.__saldo}")
        else:
            print("Saldo insuficiente")

    @property
    def nombre(self):
        print("Obteniendo el nombre de la cuenta...")
        return self.__nombre

    @property
    def saldo(self):
        print("Obteniendo saldo...")
        return self.__saldo

    @property
    def dni(self):
        return self.__dni


cuenta = CuentaBancaria("Caja ahora", 1000, 12)
cuenta.depositar(500)
cuenta.retirar(200)
```
