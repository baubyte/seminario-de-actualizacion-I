## [Pydantic](https://docs.pydantic.dev/latest/)

Pydantic es una poderosa biblioteca de validación de datos y gestión de configuraciones para Python, diseñada para mejorar la robustez y confiabilidad de su código. Desde tareas básicas, como verificar si una variable es un entero, hasta tareas más complejas, como asegurar que las claves y valores de diccionarios profundamente anidados tengan los tipos de datos correctos, Pydantic puede manejar casi cualquier escenario de validación de datos con un mínimo de código repetitivo.

Una de las principales atracciones de Python es que es un lenguaje de tipado dinámico. El tipado dinámico significa que los tipos de variables se determinan en tiempo de ejecución, a diferencia de los lenguajes de tipado estático donde se declaran explícitamente en tiempo de compilación. Si bien el tipado dinámico es excelente para el desarrollo rápido y la facilidad de uso, a menudo se necesita una verificación de tipos y una validación de datos más robustas para las aplicaciones del mundo real. Aquí es donde la biblioteca Pydantic de Python lo tiene cubierto.

Pydantic es una poderosa biblioteca de Python que aprovecha las sugerencias de tipo (type hints) para ayudarlo a validar y serializar fácilmente sus esquemas de datos. Esto hace que su código sea más robusto, legible, conciso y más fácil de depurar. Pydantic también se integra bien con muchas herramientas y IDE de tipado estático populares, lo que le permite detectar problemas de esquema antes de ejecutar su código.

Algunas de las características distintivas de Pydantic incluyen:

- Personalización: Casi no hay límite para los tipos de datos que puedes validar con Pydantic. Desde tipos primitivos de Python hasta estructuras de datos altamente anidadas, Pydantic te permite validar y serializar casi cualquier objeto de Python.
- Flexibilidad: Pydantic te da control sobre cuán estricto o flexible quieres ser al validar tus datos. En algunos casos, es posible que desee coaccionar los datos entrantes al tipo correcto. Por ejemplo, podría aceptar datos que se supone que son un flotante pero se reciben como un entero. En otros casos, es posible que desee aplicar estrictamente los tipos de datos que está recibiendo. Pydantic te permite hacer cualquiera de las dos cosas.
- Serialización: Puedes serializar y deserializar objetos Pydantic como diccionarios y cadenas JSON. Esto significa que puede convertir sin problemas tus objetos Pydantic hacia y desde JSON. Esta capacidad ha dado lugar a API auto-documentadas e integración con casi cualquier herramienta que admita esquemas JSON.
- Rendimiento: Gracias a su lógica de validación central escrita en Rust, Pydantic es excepcionalmente rápido. Esta ventaja de rendimiento le brinda un procesamiento de datos rápido y confiable, especialmente en aplicaciones de alto rendimiento, como las API REST que necesitan escalar a una gran cantidad de solicitudes.
- Ecosistema y Adopción en la Industria: Pydantic es una dependencia de muchas bibliotecas populares de Python como FastAPI, LangChain y Polars. También es utilizado por la mayoría de las empresas de tecnología más grandes y en muchas otras industrias. Este es un testimonio del soporte de la comunidad, la confiabilidad y la resistencia de Pydantic.


Estas son algunas características clave que hacen de Pydantic una atractiva biblioteca de validación de datos, y podrás verla en acción a lo largo de este tutorial.

## Usando Modelos

La forma principal de Pydantic para definir esquemas de datos es a través de modelos. Un modelo de Pydantic es un objeto, similar a una clase de datos (dataclass) de Python, que define y almacena datos sobre una entidad con campos anotados. A diferencia de las clases de datos, el enfoque de Pydantic se centra en el análisis, la validación y la serialización automática de datos.

La mejor manera de entender esto es crear sus propios modelos, y eso es lo que hará a continuación.



### Trabajando con BaseModels de Pydantic

Supongamos que estámos construyendo una aplicación utilizada por un departamento de recursos humanos para administrar la información de los empleados, y necesita una forma de verificar que la nueva información del empleado tenga el formato correcto. Por ejemplo, cada empleado debe tener una identificación, nombre, correo electrónico, fecha de nacimiento, salario, departamento y selección de beneficios.

Para definir su modelo de empleado, crea una clase que hereda de BaseModel de Pydantic


```python
from datetime import date
from uuid import UUID, uuid4
from enum import Enum
from pydantic import BaseModel, EmailStr, ValidationError


class Departamento(Enum):
    RRHH = "RRHH"
    VENTAS = "VENTAS"
    TI = "TI"
    INGENIERIA = "INGENIERIA"

class Empleado(BaseModel):
    id_empleado: UUID = uuid4()
    nombre: str
    email: EmailStr
    fecha_de_nacimiento: date
    salario: float
    departamento: Departamento
    beneficios_elegidos: bool
```

Primero, veremos las dependencias que necesitas para definir tu modelo de empleado. Luego, creamos una enumeración (`enum`) para representar los diferentes departamentos de tu empresa, y la usaremos para anotar el campo `departamento` en tu modelo de empleado.

Luego, definiremos el modelo Pydantic, `Empleado`, que hereda de `BaseModel` y define los nombres y los tipos esperados de sus campos de empleado a través de anotaciones. Aquí hay un desglose de cada campo que ha definido de `Empleado` y cómo Pydantic lo valida cuando se crea una instancia de un objeto `Empleado`:

*   `id_empleado`: Este es el UUID del empleado para el que desea almacenar información. Al usar la anotación `UUID`, Pydantic asegura que este campo sea siempre un UUID válido. A cada instancia de `Empleado` se le asignará un UUID de forma predeterminada, como especificó al llamar a `uuid4()`.
*   `nombre`: El nombre del empleado, que Pydantic espera que sea una cadena.
*   `email`: Pydantic se asegurará de que cada correo electrónico del empleado sea válido utilizando la biblioteca `email-validator` de Python en segundo plano.
*   `fecha_de_nacimiento`: La fecha de nacimiento de cada empleado debe ser una fecha válida, como se anota con `date` del módulo `datetime` de Python. Si pasa una cadena a `fecha_de_nacimiento`, Pydantic intentará analizarla y convertirla en un objeto `date`.
*   `salario`: Este es el salario del empleado y se espera que sea un flotante.
*   `departamento`: El departamento de cada empleado debe ser uno de `HR`, `VENTAS`, `TI` o `INGENIERIA`, como se define en su enumeración `Departamento`.
*   `beneficios_elegidos`: Este campo almacena si el empleado ha elegido beneficios, y Pydantic espera que sea un booleano.

La forma más sencilla de crear un objeto `Empleado` es instanciarlo como lo haría con cualquier otro objeto de Python.

Vamos a crear la instancia.



```python
empleado:Empleado = Empleado(
    nombre="SAI",
    email="lcd-sai182@ugr.edu.ar",
    fecha_de_nacimiento="1998-04-02",
    salario=123_000.00,
    departamento="TI",
    beneficios_elegidos=True,
)
print('Se creo un objeto Empleado:', empleado)
```

Se creó un objeto con todos los campos de empleado requeridos. Pydantic valida y coacciona con éxito los campos que pasó y crea un objeto `Empleado` válido. Observe cómo Pydantic convierte automáticamente tu cadena de fecha en un objeto `date` y tu cadena `TI` a su respectivo enum `Departamento`.

A continuación, vamos crear una instancia con datos no válidos.


```python
import traceback
```

```python
try:
  empleado:Empleado = Empleado(
      id_empleado="123",
      nombre=False,
      email="lcd-sai182@ugr.edu.ar",
      fecha_de_nacimiento="1939804-02",
      salario="pago máximo",
      departamento="PRODUCTO",
      beneficios_elegidos=300,)
except ValidationError as e:
    traceback.print_exc()
```

Creamos un objeto `Empleado` con campos de datos no válidos. Pydantic te proporciona un mensaje de error detallado para cada campo, indicándote lo que se esperaba, lo que se recibió y dónde puede ir para obtener más información sobre el error.

Esta validación detallada es poderosa porque le impide almacenar datos no válidos en `Empleado`. Esto también le da la confianza de que los objetos `Empleados` que instancia sin errores contienen los datos que espera, y puede confiar en estos datos en el resto de su código o en otras aplicaciones.

`BaseModel` de Pydantic está equipado con un conjunto de métodos que facilitan la creación de modelos a partir de otros objetos, como diccionarios y JSON.

Por ejemplo, si desea crear una instancia de un objeto `Empleado` a partir de un diccionario, puede utilizar el método de clase `.model_validate()`.


```python
empleado_dict:dict = {
    "nombre":"SAI",
    "email":"lcd-sai182@ugr.edu.ar",
    "fecha_de_nacimiento":"1998-04-02",
    "salario":123_000.00,
    "departamento":"TI",
    "beneficios_elegidos":True,
}
empleado:Empleado = Empleado.model_validate(empleado_dict)
print('Se creo un objeto Empleado:', empleado)
```

Creamos `empleado_dict`, un diccionario con los campos del empleado, y lo pasamos a `.model_validate()` para crear una instancia de `Empleado`. Internamente, Pydantic valida cada entrada del diccionario para asegurarse de que se ajusta a los datos que espera. Si alguno de los datos no es válido, Pydantic generará un error de la misma manera que vio anteriormente. También notificará si faltan campos en el diccionario.

Puede hacer lo mismo con objetos JSON usando `.model_validate_json()`.


```python
empleado_json:str = """{
  "nombre": "SAI",
  "email": "lcd-sai182@ugr.edu.ar",
  "fecha_de_nacimiento": "1998-04-02",
  "salario": 123000.0,
  "departamento": "TI",
  "beneficios_elegidos": true}"""
empleado:Empleado = Empleado.model_validate_json(empleado_json)
print('Se creo un objeto Empleado:', empleado)
```

En este ejemplo, `empleado_json` es una cadena JSON válida que almacena los campos de su empleado, y utiliza `.model_validate_json()` para validar y crear un objeto `Empleado` a partir de `empleado_json`. Si bien puede parecer sutil, la capacidad de crear y validar modelos Pydantic a partir de JSON es poderosa porque JSON es una de las formas más populares de transferir datos a través de la web. Esta es una de las razones por las que FastAPI confía en Pydantic para crear API REST.

También puedes serializar modelos Pydantic como diccionarios y JSON.


```python
print('El diccionario del objecto empleado es:', empleado.model_dump())
```

```python
print('El JSON del objecto empleado es:', empleado.model_dump_json())
```

Se utiliza `.model_dump()` y `.model_dump_json()` para convertir el modelo `empleado` a un diccionario y a una cadena JSON, respectivamente. Observe cómo `.model_dump_json()` devuelve un objeto JSON con `fecha_de_nacimiento` y `departamento` almacenados como cadenas.

Si bien Pydantic ya validó estos campos y convirtió su modelo a JSON, quien use este JSON posteriormente no sabrá que `fecha_de_nacimiento` debe ser una fecha válida y que `departamento` debe ser una categoría en su enumeración `Departamento`. Para resolver esto, puede crear un esquema JSON a partir del modelo `Empleado`.

Los esquemas JSON te indican qué campos se esperan y qué valores se representan en un objeto JSON. Puede pensar en esto como la versión JSON de la definición de su clase `Empleado`. A continuación, se muestra cómo generar un esquema JSON para `Empleado`:


```python
empleado_json_schema:dict = Empleado.model_json_schema()
print('El esquema JSON es:', empleado_json_schema)
```

Cuando llama a `.model_json_schema()`, se obtiene un diccionario que representa el esquema JSON del modelo. La primera entrada muestra los valores que puede tomar `departamento`. También se puede ver información sobre cómo deben formatearse sus campos. Por ejemplo, según este esquema JSON, se espera que `id_empleado` sea un UUID y que `fecha_de_nacimiento` sea una fecha.

Puedes convertir tu esquema JSON en una cadena JSON usando `json.dumps()`, lo que permite que casi cualquier lenguaje de programación valide los objetos JSON producidos por su modelo `Empleado`. En otras palabras, Pydantic no solo puede validar los datos entrantes y serializarlos como JSON, sino que también proporciona a otros lenguajes de programación la información que necesitan para validar los datos de su modelo a través de esquemas JSON.

Con esto, ahora comprendes cómo usar `BaseModel` de Pydantic para validar y serializar sus datos.


### Usando Campos para Personalización y Metadatos

Hasta ahora, el modelo `Empleado` valida el tipo de datos de cada campo y asegura que algunos de los campos, como `email`, `fecha_de_nacimiento` y `departamento`, adopten formatos válidos. Sin embargo, digamos que también desea asegurarse de que `salario` sea un número positivo, que `nombre` no sea una cadena vacía y que `email` contenga el nombre de dominio de su empresa. Para esto vamos a utilizar la clase `Field` de Pydantic.

La clase `Field` te permite personalizar y agregar metadatos a los campos de su modelo.

```python
from pydantic import Field

class Empleado(BaseModel):
    id_empleado: UUID = Field(default_factory=uuid4, frozen=True)
    nombre: str = Field(min_length=1, frozen=True)
    email: EmailStr = Field(pattern=r".+@ugr\.edu\.ar$")
    fecha_de_nacimiento: date = Field(alias="fecha_nacimiento", repr=False, frozen=True)
    salario: float = Field(alias="sueldo", gt=0, repr=False)
    departamento: Departamento
    beneficios_elegidos: bool
```

Aquí, importamos `Field` junto con las otras dependencias que usó anteriormente, y asigna valores predeterminados a algunos de los campos de `Empleados`. A continuación, se muestra un desglose de los parámetros de `Field` que se utilizó para agregar validación y metadatos adicionales a sus campos:

* **default_factory:**  Se usa para definir una función que genera valores predeterminados. En el ejemplo anterior, establece `default_factory` en `uuid4`. Esto llama a `uuid4()` para generar un UUID aleatorio para `id_empleado` cuando sea necesario. También puede usar una función lambda para mayor flexibilidad.
* **frozen:** Este es un parámetro booleano que puede establecer para hacer que sus campos sean inmutables. Esto significa que, cuando `frozen` se establece en `True`, el campo correspondiente no se puede cambiar después de crear una instancia de su modelo. En este ejemplo, `id_empleado`, `nombre` y `fecha_de_nacimiento` se vuelven inmutables usando el parámetro `frozen`.
* **min_length:** Puede controlar la longitud de los campos de cadena con `min_length` y `max_length`. En el ejemplo anterior, se asegura de que `nombre` tenga al menos un carácter.
* **pattern:** Para campos de cadena, puede establecer `pattern` a una expresión regular (`regex`) para que coincida con el patrón que espera para ese campo. Por ejemplo, cuando usa la expresión regular en el ejemplo anterior para `email`, Pydantic se asegurará de que cada correo electrónico termine con `@ugr.edu.ar`.
* **alias:** Puede usar este parámetro cuando desea asignar un alias a sus campos. Por ejemplo, puede permitir que `fecha_de_nacimiento` se llame `fecha_nacimiento` o `salario` se llame `sueldo`. Puede usar estos alias al crear una instancia o serializar un modelo.
* **gt:** Este parámetro, abreviatura de "greater than" (mayor que), se usa para campos numéricos para establecer valores mínimos. En este ejemplo, establecer `gt=0` asegura que `salario` siempre sea un número positivo. Pydantic también tiene otras restricciones numéricas, como `lt` que significa "less than" (menor que).
* **repr:** Este parámetro booleano determina si un campo se muestra en la representación de campo del modelo. En este ejemplo, no verá `fecha_de_nacimiento` o `salario` cuando imprima una instancia de `Empleado`.

Para ver esta validación adicional en acción, observe lo que sucede cuando intenta crear un modelo `Empleado` con datos incorrectos.

```python
empleado_datos_incorrectos:dict = {
     "nombre": "",
     "email": "lcd-sai182@gmail.com",
     "fecha_nacimiento": "1998-04-02",
     "salario": -10,
     "departamento": "TI",
     "beneficios_elegidos": True,
}
try:
  Empleado.model_validate(empleado_datos_incorrectos)
except ValidationError as e:
    traceback.print_exc()
```

Vamos a crear un instancia con datos incorrectos del modelo `Empleado` actualizado y se va a validar un diccionario con datos incorrectos. En respuesta, Pydantic le proporciona tres errores de validación que indican que el `nombre` debe tener al menos un carácter, el `email` debe coincidir con el nombre de dominio de su empresa y el `salario` debe ser mayor que cero.

Ahora observe las características adicionales que obtiene cuando valida los datos correctos de `Empleado`.


```python
empleado_datos_incorrectos:dict = {
     "nombre": "SAI",
     "email": "lcd-sai182@ugr.edu.ar",
     "fecha_nacimiento": "1998-04-02",
     "sueldo": 100_000,
     "departamento": "TI",
     "beneficios_elegidos": True,
}
empleado:Empleado = Empleado.model_validate(empleado_datos_incorrectos)
print('Se creo un objeto Empleado:', empleado)
```

En este bloque, crearemos un diccionario y un modelo `Empleado` con `.model_validate()`. En `empleado_datos`, observe cómo usó `fecha_nacimiento` en lugar de `fecha_de_nacimiento` y `sueldo` en lugar de `salario`. Pydantic reconoce estos alias y asigna sus valores al nombre de campo correcto internamente.

Debido a que configuró `repr=False`, puede ver que `salario` y `fecha_de_nacimiento` no se muestran en la representación de `Empleado`. Tiene que acceder a ellos explícitamente como atributos para ver sus valores. Por último, observe lo que sucede cuando intenta cambiar un campo congelado (frozen).


```python
empleado.departamento = "HR"
try:
  empleado.nombre = "Herramientas de Software para el Análisis de Datos"
except ValidationError as e:
    traceback.print_exc()
```

Aquí, primero cambia el valor de `departamento` de `TI` a `HR`. Esto es perfectamente aceptable porque `departamento` no es un campo congelado. Sin embargo, cuando intenta cambiar `nombre`, Pydantic le da un error que dice que `nombre` es un campo congelado.

Ahora tiene una comprensión sólida de las clases `BaseModel` y `Field` de Pydantic. Solo con estas, puede definir muchas reglas de validación y metadatos diferentes en sus esquemas de datos, pero a veces esto no es suficiente.

## Trabajando con Validadores

Hasta ahora, hemoss utilizado el `BaseModel` de Pydantic para validar campos de modelo con tipos predefinidos, e incorporamos `Field` para personalizar aún más la validación. Si bien podemos llegar bastante lejos solo con `BaseModel` y `Field`, para escenarios de validación más complicados que requieren lógica personalizada, necesitarás usar los `validadores` de Pydantic.

Con los validadores, puedes ejecutar casi cualquier lógica de validación que puedas expresar en una función.



### Validando Modelos y Campos

Continuando con el ejemplo del empleado, vamos a suponer que la empresa tiene una política de que solo contrata empleados que tengan al menos dieciocho años. Cada vez que creas un nuevo objeto `Empleado`, necesitamos asegurarnos de que el empleado sea mayor de dieciocho años. Para manejar esto, podrías agregar un campo `edad` y usar la clase `Field` para asegurar que el empleado tenga al menos dieciocho años. Sin embargo, esto parece redundante ya que hemos almacenado la fecha de nacimiento del empleado.

Una mejor solución es usar un `validador de campo` de Pydantic. Los validadores de campo permiten aplicar lógica de validación personalizada a los campos `BaseModel` agregando métodos de clase a tu modelo. Para asegurar que todos los empleados tengan al menos dieciocho años, podemos agregar el siguiente `validador de campo` al modelo `Empleado`.


```python
from pydantic import field_validator

class Empleado(BaseModel):
    id_empleado: UUID = Field(default_factory=uuid4, frozen=True)
    nombre: str = Field(min_length=1, frozen=True)
    email: EmailStr = Field(pattern=r".+@ugr\.edu\.ar$")
    fecha_de_nacimiento: date = Field(alias="fecha_nacimiento", repr=False, frozen=True)
    salario: float = Field(alias="sueldo", gt=0, repr=False)
    departamento: Departamento
    beneficios_elegidos: bool


    @field_validator("fecha_de_nacimiento")
    @classmethod
    def validacion_edad(cls, fecha_de_nacimiento: date) -> date:
        hoy = date.today()
        dieciocho_anios_atras = date(hoy.year - 18, hoy.month, hoy.day)

        if fecha_de_nacimiento > dieciocho_anios_atras:
            raise ValueError("El empleado debe ser mayor de 18 años.")

        return fecha_de_nacimiento
```

Importamos `field_validator` y lo usamos para decorar un método de clase en `Empleado` llamado `.validacion_edad()`. Los validadores de campo deben definirse como métodos de clase. En `.validacion_edad()`, calculamos la fecha de hoy pero hace dieciocho años. Si la `fecha_de_nacimiento` del empleado es posterior a esa fecha, se genera un error.

Veamos un ejemplo de como funciona este validador.


```python
from datetime import timedelta

empleado_menor:dict = {
     "nombre": "SAI",
     "email": "lcd-sai182@ugr.edu.ar",
     "fecha_nacimiento": date.today() - timedelta(days=365 * 17),
     "sueldo": 100_000,
     "departamento": "TI",
     "beneficios_elegidos": True,
}

try:
  empleado:Empleado = Empleado.model_validate(empleado_menor)
except ValidationError as e:
    traceback.print_exc()
```

En este ejemplo, especificamos una `fecha_de_nacimiento` que está diecisiete años atrás de la fecha actual. Cuando se llama a `.model_validate()` para validar, se obtiene un error que dice que los empleados deben tener al menos dieciocho años.

Como podemos imaginar, `field_validator()` de Pydantic te permite personalizar arbitrariamente la validación de campos. Sin embargo, `field_validator()` no funcionará si deseamos comparar múltiples campos entre sí o validar el modelo como un todo. Para esto, necesitamos usar validadores de modelo.

Como ejemplo, suponemos que la empresa solo contrata trabajadores por contrato en el departamento de TI. Debido a esto, los trabajadores de TI no califican para beneficios y el campo `beneficios_elegidos` debe ser `False`. Podemos usar `model_validator()` de Pydantic para aplicar esta restricción.


```python
import sys
if sys.version.split(".")[1] <= "10":
    from typing_extensions import Self
else:
    from typing import Self
from pydantic import model_validator


class Empleado(BaseModel):
    id_empleado: UUID = Field(default_factory=uuid4, frozen=True)
    nombre: str = Field(min_length=1, frozen=True)
    email: EmailStr = Field(pattern=r".+@ugr\.edu\.ar$")
    fecha_de_nacimiento: date = Field(alias="fecha_nacimiento", repr=False, frozen=True)
    salario: float = Field(alias="sueldo", gt=0, repr=False)
    departamento: Departamento
    beneficios_elegidos: bool


    @field_validator("fecha_de_nacimiento")
    @classmethod
    def validacion_edad(cls, fecha_de_nacimiento: date) -> date:
        hoy = date.today()
        dieciocho_anios_atras = date(hoy.year - 18, hoy.month, hoy.day)

        if fecha_de_nacimiento > dieciocho_anios_atras:
            raise ValueError("El empleado debe ser mayor de 18 años.")

        return fecha_de_nacimiento

    @model_validator(mode="after")
    def validacion_de_beneficios(self) -> Self:
        departamento:str = self.departamento
        beneficios_elegidos:bool = self.beneficios_elegidos

        if departamento == Departamento.TI and beneficios_elegidos:
            raise ValueError(
                "Los empleado de TI no aplican para los beneficios"
            )
```

Aquí, agregamos el tipo `Self` de Python y `model_validator()` de Pydantic a las importaciones. Luego, creamos un método, `.validacion_de_beneficios()`, que genera un error si el empleado pertenece al departamento de TI y el campo `beneficios_elegidos` es `True`. Cuando configuramos `mode` como `after` en `@model_validator`, Pydantic espera hasta después de que hayas instanciado el modelo para ejecutar `.validacion_de_beneficios()`.

> Nota: Es posible que hayas notado que `.validacion_de_beneficios()` está anotado con el tipo `Self` de Python. Esto se debe a que `.validacion_de_beneficios()` devuelve la instancia de la clase `Empleado`, y el tipo `Self` es la anotación preferida para esto.

Vamos a ver como funciona este validador.


```python
from datetime import timedelta

empleado_nuevo:dict = {
     "nombre": "SAI",
     "email": "lcd-sai182@ugr.edu.ar",
     "fecha_nacimiento": "1998-04-02",
     "sueldo": 100_000,
     "departamento": "TI",
     "beneficios_elegidos": True,
}

try:
  empleado:Empleado = Empleado.model_validate(empleado_nuevo)
except ValidationError as e:
    traceback.print_exc()
```

En este ejemplo, intentamos crear un modelo `Empleado` con un departamento de TI y `beneficios_elegidos` establecido en `True`. Cuando llamamos a `.model_validate()`, Pydantic genera un error que te informa que los empleados de TI no califican para beneficios.

Con los validadores de modelo y de campo, podemos implementar casi cualquier validación personalizada. Ahora deberíamos tener una base sólida para crear modelos Pydantic para tus propios casos de uso.


### Usando Decoradores de Validación para Validar Funciones

Si bien `BaseModel` es la clase principal de Pydantic para validar esquemas de datos, también podemos usar Pydantic para validar argumentos de funciones usando el decorador `@validate_call`. Esto permite crear funciones robustas con errores de tipo informativos sin tener que implementar manualmente la lógica de validación.

Para ver cómo funciona esto, vamos a suponer que estamos escribiendo una función que envía facturas a los clientes después de que han realizado una compra. Tu función recibe el nombre del cliente, el correo electrónico, los artículos comprados y el monto total de la facturación, y construye y envía un correo electrónico. Necesitamos validar todas estas entradas porque podría resultar que el correo electrónico no se envíe, tenga un formato incorrecto o que el cliente sea facturado incorrectamente.


```python
import os
import time
from typing import Annotated
from pydantic import PositiveFloat, validate_call

@validate_call
def enviar_factura(
    cliente_nombre: Annotated[str, Field(min_length=1)],
    cliente_email: EmailStr,
    articulos_comprados: list[str],
    total_facturado: PositiveFloat,
) -> str:

    email_str = f"""
Estimado {cliente_nombre}, \n
Gracias por elegir xyz inc!
Usted comprobo un total de ${total_facturado:,.2f} con los siguiente artículos: \n
{os.linesep.join(articulos_comprados)}
    """

    print(f"Enviando el correo...\n {cliente_email}...")
    time.sleep(2)

    return email_str
```

Primero, importamos las dependencias necesarias para escribir y anotar `enviar_factura()`. Luego creamos `enviar_factura()` decorada con `@validate_call`. Antes de ejecutar `enviar_factura()`, `@validate_call` se asegura que cada entrada se ajuste a las anotaciones. En este caso, `@validate_call` verifica si `cliente_nombre` tiene al menos un carácter, si `cliente_email` tiene el formato correcto, si `articulos_comprados` es una lista de str y si `total_facturado` es un flotante positivo.

Si una de las entradas no se ajusta a tu anotación, Pydantic generará un error similar a lo que ya has visto con `BaseModel`. Si todas las entradas son válidas, `enviar_factura()` crea una cadena y simula enviarla a tu cliente con `time.sleep(2)`.

> Nota: Es posible que hayas notado que `cliente_nombre` está anotado con el tipo `Annotated` de Python. En general, puedes usar `Annotated` cuando deseas proporcionar metadatos sobre un argumento de función. Pydantic recomienda usar `Annotated` cuando necesitas validar un argumento de función que tiene metadatos especificados por `Field`.

> Sin embargo, si usas `default_factory` para asignar un valor predeterminado a tu argumento de función, debes asignar el argumento directamente a una instancia de `Field`. Puedes ver un ejemplo de esto en la [documentación de Pydantic](https://docs.pydantic.dev/latest/concepts/validation_decorator/#using-field-to-describe-function-arguments).

Vamos a ver un ejemplo.

```python
try:
  email_str:str = enviar_factura(
      cliente_nombre="",
      cliente_email="lcd-sai182ugr.edu.ar",
      articulos_comprados=["torta", "masitas", 17],
      total_facturado=0,
  )
except ValidationError as e:
    traceback.print_exc()
```

En este ejemplo, pasamos argumentos de función no válidos. `@validate_call`de Pydantic reconoce esto y genera errores que indican que `cliente_nombre` necesita al menos un carácter, `cliente_email` no es válido, `articulos_comprados` debe contener cadenas y `total_facturado` debe ser mayor que cero.

Veamos que pasa cuando las entradas son válidas.

```python
try:
  email_str:str = enviar_factura(
      cliente_nombre="SAI",
      cliente_email="lcd-sai182@ugr.edu.ar",
      articulos_comprados=["torta", "masitas"],
      total_facturado=12_200,
  )
  print(email_str)
except ValidationError as e:
    traceback.print_exc()
```

Si bien `@validate_call` no es tan flexible como `BaseModel`, aún podemos usarlo para aplicar una potente validación a los argumentos de una función. Esto ahorra mucho tiempo y permite evitar escribir código repetitivo de verificación de tipos y lógica de validación. Si has hecho esto antes, sabes lo engorroso que puede ser escribir sentencias `assert` para cada uno de los argumentos de una función. Para muchos casos de uso, `@validate_call` se encarga de esto por ti.


## Gestionando la Configuración

Una de las formas más populares de configurar aplicaciones Python es con variables de entorno. Una variable de entorno es una variable que reside en el sistema operativo, fuera de tu código Python, pero que puede ser leída por tu código u otros programas. Ejemplos de datos que queremos almacenar como variables de entorno incluyen claves secretas, credenciales de bases de datos, credenciales de API, direcciones de servidores y tokens de acceso.

Las variables de entorno a menudo cambian entre el desarrollo y la producción, y muchas contienen información sensible. Debido a esto, necesitamos una forma robusta de analizar, validar e integrar variables de entorno en el código. Este es un caso de uso perfecto para `pydantic-settings`, y eso es lo que explotaremos en esta sección.



### Configurando Aplicaciones con BaseSettings

`pydantic-settings` es una de las formas más poderosas de gestionar variables de entorno en Python, y ha sido ampliamente adoptada y recomendada por bibliotecas populares como FastAPI. Podemos usar `pydantic-settings` para crear modelos, similares a `BaseModel`, que analizan y validan variables de entorno.

La clase principal en `pydantic-settings` es `BaseSettings`, y tiene todas las mismas funcionalidades que `BaseModel`. Sin embargo, si creamos un modelo que hereda de `BaseSettings`, el inicializador del modelo intentará leer cualquier campo que no se haya pasado como argumento de palabra clave desde las variables de entorno.

Para ver cómo funciona esto, vamos a suponer que la aplicación se conecta a una base de datos y a otro servicio de API. Las credenciales de la base de datos y la clave de la API pueden cambiar con el tiempo y, a menudo, cambian según el entorno en el que estás implementando. Para manejar esto, vamos a crear el siguiente modelo `BaseSettings`:


```python
from pydantic import HttpUrl
from pydantic_settings import BaseSettings

class AppConfig(BaseSettings):
    base_de_datos_host: HttpUrl
    base_de_datos_usuario: str = Field(min_length=5)
    base_de_datos_password: str = Field(min_length=10)
    api_key: str = Field(min_length=20)
```

En este script, importaremos las dependencias necesarias para crear tu modelo `BaseSettings`. Observa que importaremos `BaseSettings` desde `pydantic_settings` con un guión bajo. Luego definiremos un modelo, `AppConfig`, que hereda de `BaseSettings` y almacena campos sobre tu base de datos y la clave de la API. En este ejemplo, `base_de_datos_host` tiene que ser una URL HTTP válida, y los campos restantes tienen una restricción de longitud mínima.

A continuación, exportaremos las variables de entorno.


```python
%env BASE_DE_DATOS_HOST=https://colab.research.google.com
%env BASE_DE_DATOS_USUARIO=SAI_USUARIO
%env BASE_DE_DATOS_PASSWORD=SAI_PASSWORD
%env API_KEY=1qhm_z3IPvMk-2NKYNjUGAO-kbti07f9R#scrollTo=l8NN1Fv4WdOa
```

Ahora crearemos una instancia de la clase `AppConfig`.

```python
app_config:AppConfig = AppConfig()
print('Se creo un objeto AppConfig:', app_config)
```

Observa cómo no especificamos ningún nombre de campo cuando instancias `AppConfig`. En cambio, el modelo `BaseSettings` lee los campos de las variables de entorno que hemos configurado. Observa también que exportamos las variables de entorno en mayúsculas, pero `AppConfig` las analizó y almacenó correctamente. Esto se debe a que `BaseSettings` no distingue entre mayúsculas y minúsculas al hacer coincidir las variables de entorno con los nombres de los campos.

Ahora vamos a crear variables de entorno no válidas.


```python
%env BASE_DE_DATOS_HOST=colab.research.google.com
%env BASE_DE_DATOS_USUARIO=SAI
%env BASE_DE_DATOS_PASSWORD=PASS
%env API_KEY=1q
```

```python
try:
  app_config:AppConfig = AppConfig()
except ValidationError:
  traceback.print_exc()
```

Esta vez, cuando intentamos instanciar `AppConfig`, `pydantic-settings` genera errores que indican que `base_de_datos_host` no es una URL válida y que los campos restantes no cumplen con la restricción de longitud mínima.

Si bien este fue un ejemplo de configuración simplificado, puedes aprovechar `BaseSettings` para analizar y validar casi cualquier cosa que necesites de tus variables de entorno. Cualquier validación que puedas hacer con `BaseModel` también puedes hacerla con `BaseSettings`, incluida la validación personalizada con validadores de modelo y de campo.


### Personalizando la Configuración con SettingsConfigDict

En el ejemplo anterior, vimos un ejemplo básico de cómo crear un modelo `BaseSettings` que analiza y valida variables de entorno. Sin embargo, es posible que desees personalizar aún más el comportamiento del modelo `BaseSettings`, y puedes lograr esto con `SettingsConfigDict`.

Supongamos que no podemos exportar manualmente cada una de las variables de entorno, lo cual suele ser el caso, y necesitas leerlas desde un archivo `.env`. Deberíamos asegurarnos de que `BaseSettings` distinga entre mayúsculas y minúsculas al analizar y que no haya variables de entorno adicionales en el archivo `.env` aparte de las que especificas del modelo.


```python
from pydantic_settings import SettingsConfigDict

class AppConfigEnv(BaseSettings):
    model_config = SettingsConfigDict(
        env_file=".env",
        env_file_encoding="utf-8",
        case_sensitive=True,
        extra="forbid",
    )

    base_de_datos_host: HttpUrl
    base_de_datos_usuario: str = Field(min_length=5)
    base_de_datos_password: str = Field(min_length=10)
    api_key: str = Field(min_length=20)
```

Este script es igual al del ejemplo anterior, excepto que esta vez vamos a  importar `SettingsConfigDict` y lo vamos a inicializar dentro de `AppConfig`. Dentro de `SettingsConfigDict`, especificamos que las variables de entorno deben leerse de un archivo llamado `.env`, se debe aplicar la distinción entre mayúsculas y minúsculas y se prohíben las variables de entorno adicionales en el archivo `.env`.

A continuación, crearemos el archivo llamado `.env` en el mismo directorio que se encuentra el archivo '.py' de ejecución y rellénalo con las siguientes variables de entorno.


```python
%%writefile .env
base_de_datos_host=https://colab.research.google.com
base_de_datos_usuario=SAI_USUARIO
base_de_datos_password=SAI_PASSWORD
api_key=1qhm_z3IPvMk-2NKYNjUGAO-kbti07f9R#scrollTo=l8NN1Fv4WdOa
```

Ahora creamos una instancia de `AppConfigEnv`.

```python
app_config_env:AppConfigEnv = AppConfigEnv()
print('Se creo un objeto AppConfigEnv:', app_config_env)
```

Ahora agregaremos algunas variables no válidas.

```python
%%writefile .env
BASE_DE_DATOS_HOST=https://colab.research.google.com
base_de_datos_usuario=SAI_USUARIO
base_de_datos_password=SAI_PASSWORD
api_key=1qhm_z3IPvMk-2NKYNjUGAO-kbti07f9R#scrollTo=l8NN1Fv4WdOa
base_de_datos_puerto=3306
```

Aquí, hemos cambiado `base_de_datos_host` a `BASE_DE_DATOS_HOST`, violando la restricción de distinción entre mayúsculas y minúsculas, y hemos agregado variables de entorno adicionales que no deberían estar allí ('base_de_datos_puerto'). Así es como responde el modelo cuando intenta validarlo


```python
try:
    app_config_env:AppConfigEnv = AppConfigEnv()
except ValidationError:
  traceback.print_exc()
```

Se obtienen una buena lista de errores que indican que falta `base_de_datos_host` y que tienes variables de entorno adicionales en tu archivo `.env`. Observa que, debido a la restricción de distinción entre mayúsculas y minúsculas, tu modelo piensa que `BASE_DE_DATOS__HOST` es una variable adicional junto con `base_de_datos_puerto`.

Hay mucho más que puedes hacer con `SettingsConfigDict` y `BaseSettings` en general, pero estos ejemplos deberían darte una idea de cómo puedes usar `pydantic-settings` para gestionar tus variables de entorno para tu propio caso de uso.
